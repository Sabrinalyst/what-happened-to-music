```python
# Project Setup
 
from pathlib import Path
import pandas as pd
```

```python
# Load Dataset

project_path = Path.cwd().parent
raw_path = project_path /'data' / 'raw'
chart_file = raw_path / 'top_100_songs_1952_to_2024.xlsx'

excel = pd.ExcelFile(chart_file)
```

```python
# Initial Dataset Validation

print(f'Number of worksheets: {len(excel.sheet_names)}')

print(f'\nFirst five worksheets:\n{excel.sheet_names[:5]}')

print(f'\nLast five worksheets:\n{excel.sheet_names[-5:]}')
```

### Project Starting Point – Observations

- Weekly charts contain exactly 100 entries.
- Column names are consistent with the historical dataset.
- No missing values are present.
- The first complete Top 100 chart (1983) is selected as the project starting point.

```python
# Select Relevant Worksheets

selected_sheets = [
    sheet
    for sheet in excel.sheet_names
    if sheet >= "Week 19830107"
]

len(selected_sheets)
```

```python
print(f'Selected worksheets: {len(selected_sheets)}')

print(f'\nFirst five:\n{selected_sheets[:5]}')

print(f'\nLast five:\n{selected_sheets[-5:]}')
```

```python
# Import Weekly Charts

weekly_charts = []

for sheet in selected_sheets:
    week_df = excel.parse(sheet)
    weekly_charts.append(week_df)
```

```python
len(weekly_charts)
```

```python
# Create Combined Dataset

chart_history = pd.concat(
    weekly_charts,
    ignore_index=True
)
```

```python
chart_history.shape
```

## Initial Data Validation

After importing all weekly UK chart sheets (1983–2024), several validation checks were performed:

- Verified the number of imported weeks.
- Checked the consistency of column names.
- Verified the number of chart entries per week.

```python
songs_per_week = chart_history.groupby('Week').size()

songs_per_week.describe()
```

```python
songs_per_week.value_counts().sort_index()
```

```python
songs_per_week[songs_per_week > 100]
```

### Data validation findings

- Most weeks contain exactly 100 chart entries.
- 140 early weeks contain 75 entries (expected historical format).
- Six weeks around the New Year contain duplicated chart positions.
- These weeks require special handling before final analysis.

```python
week_1993 =chart_history[chart_history["Week"] == "26 December 1993- 1 January 1994"]
week_1993.head(15)
```

```python
display(week_1993.shape)
week_1993["Position"].value_counts().sort_index()
```

```python
# Remove Exact Duplicates

week_1993 = chart_history[chart_history['Week'] == '26 December 1993- 1 January 1994']
print(f'Number of duplicates in week "26 December 1993- 1 January 1994": {week_1993.duplicated().sum()}')
```

```python
week_1993 = week_1993.drop_duplicates()
week_1993.shape
```

```python
# Remove Exact Duplicates

chart_history_clean = chart_history.drop_duplicates()

print(f'Raw dataset: {chart_history.shape}')
print(f'Clean dataset: {chart_history_clean.shape}')
```

```python
# Validate Weekly Chart Sizes

songs_per_week_clean = chart_history_clean.groupby('Week').size()

songs_per_week_clean.value_counts().sort_index()
```

```python
# Validate Top 50 Completeness

top50_per_week = (
    chart_history_clean[
        chart_history_clean['Position'] <= 50
    ]
    .groupby('Week')
    .size()
)

top50_per_week.value_counts().sort_index()
```

```python
# Identify Remaining Top 50 Anomalies

incomplete_top50_weeks = top50_per_week[top50_per_week < 50].index

for week in incomplete_top50_weeks:
    week_df = chart_history_clean[
        chart_history_clean['Week'] == week
    ]

    missing_positions = set(range(1, 51)) - set(week_df['Position'])

    print(f'{week}\nMissing position(s): {sorted(missing_positions)}\n'
)
```

## Validation Conclusion

The UK chart dataset has been successfully validated and cleaned for further analysis.

Key validation results:

- Exact duplicate records were identified and removed.
- Historical differences in weekly chart size were confirmed and documented.
- The cleaned Top 50 dataset is complete for 2,143 of 2,146 weeks.
- Three weeks contain a single missing Top 50 entry (positions 42, 43 and 46).

The remaining anomalies will be verified against the Official Charts Company and documented separately before being integrated into the final dataset.

Overall, the cleaned dataset provides a reliable and transparent foundation for the next stage of the project.
