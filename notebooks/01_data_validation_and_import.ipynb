```python
selected_sheets = [
    sheet
    for sheet in excel.sheet_names
    if sheet >= "Week 19830107"
]

len(selected_sheets)
```

```python
selected_sheets[:5]
```

```python
selected_sheets[-5:]
```

```python
test_charts = []

for sheet in selected_sheets[:5]:
    week_df = excel.parse(sheet)

    test_charts.append(week_df)
```

```python
len(test_charts)
```

```python
type(test_charts[0])
```

```python
weekly_charts = []

for sheet in selected_sheets:
    week_df = excel.parse(sheet)
    weekly_charts.append(week_df)
```

```python
len(weekly_charts)
```

```python
chart_history = pd.concat(
    weekly_charts,
    ignore_index=True
)
```

```python
chart_history.shape
```

## Initial Data Validation

After importing all weekly UK chart sheets (1983–2024), several validation checks were performed:

- Verified the number of imported weeks.
- Checked the consistency of column names.
- Compared the number of chart entries per week.

```python
chart_history.groupby("Week").size().describe()
```

```python
chart_history.groupby("Week").size().value_counts().sort_index()
```

```python
songs_per_week = chart_history.groupby("Week").size()

songs_per_week[songs_per_week > 100]
```

### Data validation findings

- Most weeks contain exactly 100 chart entries.
- 140 early weeks contain 75 entries (expected historical format).
- Six weeks around the New Year contain duplicated chart positions.
- These weeks require special handling before final analysis.
